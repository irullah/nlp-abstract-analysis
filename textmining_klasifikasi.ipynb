Preprocessing Data

```python
import nltk
import string
import re
```

```python
def case_folding(abstrak):
    abstrak = abstrak.lower()
    return abstrak
df['abstrak_lower'] = df['Abstrak'].apply(lambda x: case_folding(x))
```

```python
df.head()
```

```python
#import stopword
from nltk.corpus import stopwords 
nltk.download('stopwords')
stopwords_indonesia = stopwords.words('indonesian')
```

```python
#install and import pysastrawi
!pip install PySastrawi
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
factory = StemmerFactory()
stemmer = factory.create_stemmer()
```

```python
def cleaning(abstrak):
    symbols = ",!\"#$%&()*+-.…/:;<=>?@[\]^_`{|}~\n"
    abstrak = re.sub(r"\s—\s", "", abstrak)
    abstrak = re.sub(r"http\S+", "", abstrak)
    for i in symbols:
        abstrak = np.char.replace(abstrak, i, ' ')
    return abstrak
df['abstrak_clean'] = df['abstrak_lower'].apply(lambda x: cleaning(x))
```

```python
df.head()
```

```python
def split(abstrak):
    return abstrak.split()
df['abstrak_split'] = df['abstrak_clean'].apply(lambda x: split(x))
```

```python
df.head()
```

```python
def stemming(abstrak):
  abstraks_clean = []    
  for word in abstrak:
      word = stemmer.stem(word)
      abstraks_clean.append(word)
  return abstraks_clean
df['abstrak_stem'] = df['abstrak_split'].apply(lambda x: stemming(x))
```

```python
df.head()
```

```python
abstrak = df['Abstrak']
X = df['abstrak_stem']
Y = df['Kategori'].values 
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
# Inisialisasi objek TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()

# Terapkan TfidfVectorizer pada dokumen
tfidf_matrix = tfidf_vectorizer.fit_transform(abstrak)

# Ubah matriks TF-IDF menjadi dataframe Pandas
df_tfidf = pd.DataFrame(
    tfidf_matrix.toarray(), columns=tfidf_vectorizer.get_feature_names_out()
)

X = df_tfidf.values
print(df_tfidf)
```

```python
X_train, X_test, y_train, y_test = train_test_split( 
    X, Y, test_size = 0.1, random_state = 100)
```

```python
knn = KNeighborsClassifier(n_neighbors = 3)
knn.fit(X_train,y_train)
y_pred=knn.predict(X_test)
```

```python
print (" Accuracy : ", accuracy_score(y_test,y_pred))
```

```python
pca = PCA(n_components =30)

X_train = pca.fit_transform(X_train)
X_test = pca.transform(X_test)

pca.explained_variance_ratio_
```
